```python
# 3. Check Subset and Superset
# Given two sets, write a program to check whether the first set is a subset of the second set and whether the second set is a superset of the first.
lis1 = list()
lis2 = list()
while True:
    user1 = input("Enter for First set (or type 'Exit'): ")
    if user1.lower() == 'exit':
        break
    lis1.append(int(user1))
while True:
    user2 = input("Enter for Second set (or type 'Exit'): ")
    if user2.lower() == 'exit':
        break
    lis2.append(int(user2))

i = 0
n1 = len(lis1)
n2 = len(lis2)
count = 0

while i < n1:
    j = 0
    found = False

    while j < n2:
        if lis1[i] == lis2[j]:
            found = True
            break
        j += 1

    if found:
        count += 1
    i += 1
set_1 = set(lis1)
set_2 = set(lis2)
if count == len(set_1):
    print("First set is a subset of the second set, and the second set is a superset of the first!")
    print("Set 1:", set_1)
    print("Set 2:", set_2)
else:
    print("First set is NOT a subset, and the second set is NOT a superset.")
    print("Set 1:", set_1)
    print("Set 2:", set_2)
```

```python
# 2. Frequency of Words
# Write a Python program to accept a sentence and create a dictionary containing the frequency of each word.
# Example: "python is easy and python is useful"
# Expected result: {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'useful': 1}
sentence_word_frequency = dict()
sentence = input("Enter a sentence: ")
sentence = sentence.lower()
words = sentence.split()
n = len(words)
i = 0
while i < n:
    word = words[i]
    count = 0
    j = 0
    while j < n:
        if words[j] == word:
            count += 1
        j += 1
    sentence_word_frequency[word] = count
    i += 1

print("\n--- Word Frequencies ---")
for word, freq in sentence_word_frequency.items():
    print(f"Word --> '{word}' | Frequency --> {freq}")
```

```python
# 3. Merge Two Dictionaries and Add Common Values
# Given two dictionaries containing item names and quantities, merge them into a single dictionary. If an item appears in both dictionaries, add its quantities.
# Example:
# d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
# d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}
# Expected result: {"Pen": 14, "Pencil": 12, "Book": 3, "Eraser": 6}
d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}
merged_dict = {}
for key in d1:
    merged_dict[key] = d1[key]
for key in d2:
    if key in merged_dict:
        merged_dict[key] += d2[key]
    else:
        merged_dict[key] = d2[key]

print("Merged Dictionary:", merged_dict)
```
