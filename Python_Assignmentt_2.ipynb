```python
#1. Write a Python program with separate functions for addition, subtraction, multiplication, and division.
# Take two numbers from the user and display the result.

def check_and_answer(ch,p,p1):
  match ch:
    case 1:
     return subtration(p,p1)
    case 2:
      return addition(p,p1)
    case 3:
      return multiplication(p,p1)
    case 4:
      return division(p,p1)
def subtration(p,p1):
  greater_value=p if(p>p1) else p1
  smaller_value=p if(p<p1) else p1
  result=greater_value-smaller_value
  return result
def addition(p,p1):
  result=p+p1
  return result
def multiplication(p,p1):
  result=p*p1
  return result
def division (p,p1):
  result=p/p1
  return result
p=(int)(input("Enter 1 st number:"))
p1=(int)(input("Enter 2 nd number:"))
ch=(int)(input ("Enter your choice: 1. Subtraction 2. Addition 3. Multiplication 4. Division:"))
result=check_and_answer(ch,p,p1)
print(result)
```

```python
 # Create functions to convert:Celsius → Fahrenheit, Fahrenheit → Celsius and Celsius → Kelvin.
def Celsius_to_Fahrenheit(user_input):
  result=(user_input*1.8)+32.0
  return result
def Fahrenheit_to_Celsius(user_input):
  result=user_input-32.0
  result=result/1.8
  return result
def Celsius_to_Kelvien(user_input):
  result=user_input+273.0
  return result
def check_and_answer(choice,user_input):
  match choice:
    case 1:
     return Celsius_to_Fahrenheit(user_input)
    case 2:
      return Fahrenheit_to_Celsius(user_input)
    case 3:
      return Celsius_to_Kelvien(user_input)
user_input=(float)(input("Enter the temperature :"))
choice=(int)(input ("Enter your choice: 1.Celsius_to_Fahrenheit 2. Fahrenheit_to_Celsius 3. Celsius_to_Kelvien:"))
result=check_and_answer(choice,user_input)
print(result)
```
